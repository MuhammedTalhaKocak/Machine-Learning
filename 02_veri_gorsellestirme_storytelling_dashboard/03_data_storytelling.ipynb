# Data Storytelling Nedir?
Veri Hikayeleştirme (Data Storytelling), analiz edilen verilerin insanların kolayca anlayabileceği grafik ve anlatımlarla sunulup, sadece sonucu aktarmakla kalmayıp izleyicinin somut bir aksiyon almasını sağlayan süreçtir.

### 3 Katmanlı Yapı
1. **Bağlam (Context):** Neden bu konuya bakıyoruz? (İş problemi/motivasyon)
2. **Bulgu (Insight):** Veri bize ne söylüyor?
3. **Aksiyon (Action):** Bu bilgiyle ne yapmalıyız?

Çoğu junior analist sadece Bulgu katmanını sunar (grafik + sayı), ama işverenler asıl Aksiyon katmanını duymak ister bağlamsız bulgu unutulur, aksiyonsuz bulgu kullanılmaz.

### Görsel Storytelling İlkeleri
1. Tek grafik, tek mesaj bir grafiğe çok fazla bilgi sıkıştırmaya çalışmayın.
2. Başlık sonucu söylemeli (örn. "Kasım'da Satışlar %30 Arttı", "Aylık Satışlar" değil).
3. Renk ile vurgu yapıp, önemli veri noktasını öne çıkarmak, gerisini soluklaştmak.
4. Tek sayıdan ziyade karşılaştırma sunmak "50.000 TL" değil, "50.000 TL, geçen aya göre %15 artış".

Sigara içen müşterilerin ortalama bahşişi, içmeyenlere göre nasıl farklılaşıyor?

```python
import pandas as pd
import seaborn as sns
df = sns.load_dataset("tips")
fark = round(df.groupby('smoker')['tip'].mean().diff().iloc[-1], 2)
print(fark)
```

Bağlam: Restoran yönetimi, garsonların motivasyonunu ve gelirini artırmak istiyor; bunun için bahşiş miktarını hangi faktörlerin etkilediğini bilmeye ihtiyaç duyuyor.

Bulgu: Sigara içenlerin bıraktıkları bahşiş ile sigara içmeyenlerin bıraktıkları bahşiş arasında yaklaşık olarak 0.02 fark vardır.

Aksiyon: Sigara içenler ve içmeyen insanların bıraktıkları bahşiş arasındaki farkın neredeyse hiç olmasından dolayı gruplara özel aksiyon alınmamasına karar verilmiştir.
