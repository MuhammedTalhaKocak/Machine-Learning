İnteraktif görselleştirme, matplotlib ve seaborn'un ürettiği statik grafiklerin yetersiz olduğu durumlarda özellikle kullanıcının grafiği yakınlaştırma, belirli bir bölgeyi seçme veya belirli bir değeri görme gibi interaktif işlemlerde kullanılır. Plotly kütüphanesi, interaktif grafikler üretmede sıkça kullanılan bir Python kütüphanesidir.

```python
import pandas as pd
import plotly.express as px
import seaborn as sns
df = sns.load_dataset("tips") # Seaborn kütüphanesinin hazır veri setlerinden olan tips veri setini kullanacağız.
fig = px.scatter(data_frame=df, x="total_bill", y="tip", color="day")
fig.show()
```

```python
veri = df.groupby("day")["tip"].mean().to_frame().reset_index() # Plotly kütüphanesi veriyi DataFrame formatında beklediği için önce veriyi DataFrame'e dönüştürdük, ardından reset_index() ile indeks değerlerini 0'dan başlattık.
fig1 = px.bar(data_frame=veri, x="day", y="tip")
fig1.show()
```

Yukarıda oluşturduğumuz interaktif grafik sayesinde verilerin üzerine geldiğimizde değerlerini ve isimlerini rahatça görebiliyoruz. Ayrıca zoom in - zoom out seçeneğiyle dilediğimiz yeri büyültüp küçültebiliriz.
