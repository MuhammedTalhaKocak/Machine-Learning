### 8) Backpropagation Nedir?
Backpropagation, büyük sinir ağlarında **Chain Rule** işlemini, çıkıştan (Hata) girişe (ağırlıklar) doğru geriye doğru sistematik olarak uygulama işlemine verilen isimdir.

#### Backpropagation Uygulama

```python
import numpy as np
x = 2
y = 10
w = 3
learning_rate = 0.01

for i in range(50):
    y_hat = w * x
    hata = (y_hat - y) ** 2
    dHata_dyhat = 2 * (y_hat - y)
    dyhat_dw = x
    dHata_dw = dHata_dyhat * dyhat_dw
    w = w - (learning_rate * dHata_dw)
    print(f"Adım {i+1}: w={w:.4f}, Hata={hata:.4f}")
```

```python
x = 4   # (girdi)
y = 20  # (gerçek/hedef değer)
w = 2   # (başlangıç ağırlığı)
learning_rate = 0.01
for i in range(50):
    tahmin = x * w  # (ağırlık * katsayı)
    hata = (tahmin-y) ** 2
    dhata_dy = 2 * (tahmin-y)
    dy_dw = x
    dhata_dw = dhata_dy * dy_dw
    w = w - (learning_rate * dhata_dw) # w_eski - (learning rate * türev)
    print(f"Adım {i+1}: w={w:.4f}, Hata={hata:.4f}")
```

```python
x = 10
y = 25
w = 1
learning_rate = 0.001
for i in range(50):
    tahmin = x * w
    hata = (tahmin - y) ** 2
    dis_turev = 2 * (tahmin - y) # Hata Fonksiyonu (tahmin - gerçek değer) ** 2 "Hatanın, Tahmine göre kısmi türevi", Eğer ben Tahmin değerini azıcık değiştirirsem, Hata fonksiyonu buna nasıl bir tepki verir (ne kadar değişir)?
    ic_turev =  x # tahmin = w * x "Tahminin, Ağırlığa göre kısmi türevi", Eğer ben yapay zekanın Ağırlık düğmesini azıcık çevirirsem, ortaya çıkan Tahmin değeri ne kadar değişir?
    hata_agirlik_turevi = dis_turev * ic_turev
    w = w - (learning_rate * hata_agirlik_turevi)
    print(f"Adım {i+1}: w={w:.4f}, Hata={hata:.4f}")
```

### Öğrenme Hızı Üzerine Gözlem
Üç örnekte de learning_rate sabit (0.01) kalmasına rağmen, x değeri büyüdükçe (2 → 4 → 10) gradyan büyüklüğü de artıyor, çünkü formülde x doğrudan çarpan olarak yer alıyor (∂Hata/∂w = 2(ŷ-y) × x). 
Bu da w'nin her adımda daha büyük adımlarla güncellenmesine yol açıyor. Bu durum öğrenmeyi hızlandırıyormuş gibi görünse de aslında riskli bir durum: adımlar çok büyürse w, minimum noktayı (y/x) atlayabilir (overshoot), hatta bazı durumlarda ıraksayabilir (diverge).
Gerçek ML projelerinde bu problemi önlemek için feature scaling (StandardScaler, MinMaxScaler gibi) kullanılır — böylece tüm özellikler benzer ölçeğe getirilir ve gradyanlar kontrol altında kalır.
