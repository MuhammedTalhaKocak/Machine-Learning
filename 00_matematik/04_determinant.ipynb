### 4) Determinant Nedir?
Determinant, bir matrisin tersinin (inverse matrix) alınabilmesi için bir koşuldur bir başka deyişle determinant, kare bir matrisin tek bir sayıya indirgenmiş halidir ve bu sayı bize matris hakkında kritik bir bilgi verir: matrisin tersinin olup olmadığını.

### Kural)
- Determinant ≠ 0 → matrisin tersi vardır
- Determinant = 0 → matrisin tersi yoktur (buna "singular matrix" / tekil matris denir)

### 2x2'lik bir matris için determinant hesaplama formülü
- | a  b |
- | c  d |

det(A) = (a × d) − (b × c)

### ML ile Bağlantısı
Normal Equation'da geçen (XᵀX)⁻¹ ifadesinde, eğer XᵀX matrisinin determinantı 0 çıkarsa bu matrisin tersi alınamaz ve regresyon denklemi çözülemez. Bu genelde veri setindeki özelliklerin birbirine çok benzemesi (multicollinearity) durumunda yaşanır.

```python
import numpy as np
B = np.array([
    [6,3],
    [2,1]
])
def determinant(B: np.ndarray) ->float: 
    a, d = B[0][0], B[1][1] 
    b, c = B[0][1], B[1][0] 
    det= (a*d) - (b*c)
    return det
det = determinant(B)
if det != 0:
    print(f"B matrisinin tersi alınabilir. Determinant: {det}")
else:
    print(f"B matrisinin tersi alınamaz. Determinant: {det}")
b_det = np.linalg.det(B) # determinant alma
print(np.isclose(b_det, 0)) # isclose fonksiyonu iki sayının "pratik olarak eşit" olup olmadığını (küçük bir tolerans payıyla) kontrol eder, floating point hatalarıyla uğraşırken bilimsel/mühendislik kodlarında standart yöntem budur, == 0 gibi direkt eşitlik kontrolünden çok daha güvenilirdir.
```

```python
ornek = np.array([
    [3,7],
    [5,9]
])
ornek_det = np.linalg.det(ornek)
print(f"Örnek Matris'inin determinantı: {ornek_det}")
ornek_ters = np.linalg.inv(ornek)
print(f"Örnek Matrisinin Tersi: {ornek_ters}")
```
