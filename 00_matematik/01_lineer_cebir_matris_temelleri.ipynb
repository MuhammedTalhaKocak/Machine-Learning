### Matris Nedir?
Matrisler satır ve sütun biçimindeki yapılardır. Her bir satır bir gözlemi, her bir sütun ise bir özelliği (bağımsız değişkeni) temsil eder.

### Dot Product Nedir?
Dot Product, Makine Öğrenmesi'nde her bir özelliğin o özelliğe ait olan katsayı ile çarpılmasıdır.
Örneğin: Bir evin fiyatı = 1000 * mkare + 300 * oda olsun. Bu örnekte bir metrekare artışında evin fiyatı 1000 TL artarken, bir oda sayısı artışında ev fiyatı 300 TL artmaktadır. Bu örnekte, 1000 ve 300 rakamları ilgili özelliklere ait katsayılardır.
Burada dikkat edilmesi gereken nokta, bir evin özelliklerini tutan matris ile bu özelliklere ait olan katsayıların çarpılıp ev fiyatlarının hesaplanabilmesi için Evin özelliklerini tutan matrisin sütun sayısı, katsayıları tutan matrisin satır sayısının eşit olması gerektiği. 

### Bias Nedir?
Bias, tüm özellikler (mkare, oda) sıfır olduğunda modelin tahmin edeceği taban değerdir. Ev örneğinde bu, arsa payı,hisse gibi özelliklerden bağımsız sabit bir maliyeti temsil eder.

```python
import numpy as np
gozlemler = np.array([
    [100,2], # 1.Gözlem
    [150,3], # 2.Gözlem
    [200,4]  # 3.Gözlem 
])
katsayilar = np.array([
    [1000], # Metrekare ağırlığı
    [300]  #  Oda sayısı ağırlığı
])
print(gozlemler.shape)
print(katsayilar.shape)
```

Yukarıda söylediğimiz gibi 1. matrisin sütun sayısı 2. matrisin satır sayısına eşit olduğundan bu iki matrisi çarpabiliriz.

```python
sonuc_matrisi = gozlemler @ katsayilar # NumPy'da 2 matrisi çarpmak için '@' işareti veya dot fonksiyonu kullanılır. Biz kolaylık olması açısından '@' işaretini kullandık.
print(sonuc_matrisi)
print(f"Sonuç Matrisinin Satır Sayısı: {sonuc_matrisi.shape[0]}, Sonuç Matrisinin Sütun Sayısı: {sonuc_matrisi.shape[1]}")
```

- 1.Ev: 100600 TL,
- 2.Ev: 150900 TL,
- 3.Ev: 201200 TL

```python
sabit = 5000
nihai_fiyat_tahminleri = sonuc_matrisi + sabit
print(nihai_fiyat_tahminleri)
```

Sabit fiyat (bias) ekledikten sonra evin fiyatı:
- 1.Ev: 105600 TL,
- 2.Ev: 155900 TL,
- 3.Ev: 206200 TL
