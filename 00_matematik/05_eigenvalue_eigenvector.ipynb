### 5)Eigenvalue ve Eigenvector Nedir?
Eigenvector, bir matrisle çarpıldığında kendi yönü değişmeyen, sadece büyüklüğü (uzunluğu) değişen özel bir vektördür. Ne kadar büyüdüğünü/küçüldüğünü gösteren çarpana da eigenvalue denir.

### Formül
- A · v = λ · v
- A = matris
- v = eigenvector (özvektör)
- λ (lambda) = eigenvalue (özdeğer)

```python
import numpy as np
A = np.array([
    [4,2],
    [1,3]
])
ozdegerler, ozvektorler = np.linalg.eig(A) # Pythonda Eigenvalue ve Eigenvector hesaplamak için NumPy kütüphanesinden linalg modülünden eig fonksiyonu kullanılır. Bu fonksiyon, sırayla verilen matrisin özdeğerlerini ve özvektörlerini döndürür.
print(f"Özdeğerler: {ozdegerler[0]}")
print(f"Özvektörler:{ozvektorler[:,0]}")
print(f"Özdeğerler: {ozdegerler[1]}")
print(f"Özvektörler:{ozvektorler[:,1]}")
```

```python
v1 = A @ ozvektorler[:,0] # birinci özvektör ile A matrisinin çarpımı
print(f"v1: {v1}\n") #   
print(f"{v1/5}", ozvektorler[:,0]) # görüldüğü üzere v1, birinci özvektörün 5 katıdır. Bu da özdeğerin 5 olduğunu gösterir.
```

 Yukarıda görüldüğü üzere özvektörün ilkini A matrisi ile çarptığımızda, vektörün yönü değişmedi sadece büyüklüğü eigenvalue (5) kadar arttı.

```python
sol_taraf = A @ ozvektorler[:,0] # burada formülün sol tarafını hesapladık.
sag_taraf = ozdegerler[0] * ozvektorler[:,0] # burada formülün sağ tarafını hesapladık.
print(np.allclose(sol_taraf, sag_taraf)) # allclose fonksiyonu formüldeki sol ve sağ tarafın birbirine tamamen eşit olup olmadığını kontrol eder.
```
