### Ters Matris (Inverse Matrix) Nedir?
Bir matrisin tersi (Inverse), o matrisle çarpıldığında birim matris (identity matrix) denen özel bir matrisi verir.
Yani: A × A⁻¹ = I
Birim matris köşegenleri 1, kalan yerleri ise 0'dan oluşur.

```python
import numpy as np
gozlemler = np.array([
    [100,3], # 1.Gözlem
    [80,2]   # 2.Gözlem
])
ters_matris_gozlemler = np.linalg.inv(gozlemler) # NumPy'da bir matrisin tersini almak için linalg modülünün inv fonksiyonu kullanılır.
birim_matris = gozlemler @ ters_matris_gozlemler # gozlemler matrisi ile bu matrisin tersinin çarpımının birim matris olacağını kontrol ediyoruz.
print(f"Birim matris:\n {birim_matris}")
```

### Sonuç
Görüldüğü üzere gozlemler matrisi ile bu matrisin tersinin çarpımı birim matris oluşturmaktadır.

### Not: Neden Bazen Tam 0 veya 1 Görmeyebiliriz?
Bilgisayarlar ondalıklı sayıları ikili (binary) sistemde, sonlu bir hassasiyetle saklar. Bu yüzden ters matris hesaplamaları gibi işlemlerde bazen köşegende tam olarak `1` yerine `0.9999999998`, sıfır olması gereken yerlerde ise tam `0` yerine `-0.0000000001` gibi çok küçük sapmalar görülebilir. Buna **floating point precision (kayan nokta hassasiyeti) hatası** denir. Bu bir hesaplama yanlışlığı değildir, bilgisayarların sayıları temsil etme biçiminden kaynaklanan doğal bir durumdur. Bizim örneğimizde sonuç tesadüfen tam `1` ve `0` çıktı, ancak daha büyük veya karmaşık matrislerde bu tür ufak sapmalarla karşılaşılması normaldir.
