### Transpose (Devrik Matris) Nedir?
2 matrisin çarpılabilmesi için 1.matrisin sütun sayısı 2.matrisin satır sayısına eşit olmalıdır ancak bazı durumlarda elimizdeki matrisler bu kurala uymaz ama matematiksel olarak çarpmamız gerekir. Burada Transpose matris devreye girer. Bir matrisin transpose matrisi o matrisin satırlarının sütun, sütunlarının satır olması dönüşümüyle elde edilir.

```python
import numpy as np
gozlemler = np.array([
    [100,3], # 1.Gözlem
    [80,2],  # 2.Gözlem
    [50,1]   # 3.Gözlem
])
gozlemler_transpose = gozlemler.T # NumPy kütüphanesinde bir matrisin transpozunu alabilmek için T öz niteliği veya transpose fonksiyonu kullanılır.
print(f"Orjinal matrisin boyutu: {gozlemler.shape}")
print(f"Transpoze edilmiş matrisin boyutu: {gozlemler_transpose.shape}")
```

Orjinal Matris 3 satır 2 sütundan oluşuyordu, transpoze edilen matris 2 satır 3 sütundan oluşuyor.

```python
print(gozlemler_transpose)
```

### Sonuç
Transpoze edilen matrisde her bir sütun bir gözlemi temsil ediyor. Bu matrisin her bir satır ise özelliklerin değerlerini temsil etmektedir.
