### Türev
Bir modelin öğrenmesi aslında hata fonksiyonunu minimize etmeye çalışmaktan geçer. Bu noktada, hatanın hangi yönde azaldığını bilmemiz gerekir, türev tam bu noktada devreye girer.
Bir fonksiyonun türevi, o fonksiyonun belirli bir noktadaki değişim hızını (eğimini) gösterir.

### Örnek
f(x) = x²  için türev:  f'(x) = 2x  (Üssü çarpan konumuna al, üssü 1 azalt.)

### Gradient Descent Algoritması Nedir?
Bu algoritma türev üzerine kuruludur. Model, hata fonksiyonunun türevini (gradyanını) hesaplayıp, o gradyanın tersi yönünde küçük adımlar atarak hatayı azaltır.

### Gradient Descent Formülü

- x_yeni = x_eski − (learning_rate × türev)
- learning_rate (η, öğrenme oranı) → adımın ne kadar büyük atılacağını kontrol eden bir sayı (genelde 0.001 - 0.1 arası küçük bir değer)
- Eğim büyükse → x_yeni daha çok değişir (büyük adım)
- Eğim sıfıra yaklaştıkça → adımlar doğal olarak küçülür, model "yavaşlar" ve minimum noktaya yumuşak iniş yapar


```python
def turev(x):
    return 2*x
x = 5
learning_rate = 0.1
for i in range(20):
    grad = turev(x)
    print(f"x={x} noktasındaki türev: {grad}")
    x_yeni = x - (learning_rate * grad)  
    print(f"Yeni x değeri: {x_yeni}")
    x = x_yeni
```

* Yukarıdaki örnekte gradient descent'i 20 adımda simüle ederek türevin gittikçe 0'a yaklaştığını, yani düzlüğe doğru indiğimizi görüyoruz. Türev küçüldükçe (x - x_yeni) farkı da küçülüyor, yani atılan adımlar otomatik olarak küçülüyor. Bu sayede model minimum noktaya yaklaştıkça "yavaşlıyor" ve noktayı atlamadan (overshoot etmeden) iniyor.
