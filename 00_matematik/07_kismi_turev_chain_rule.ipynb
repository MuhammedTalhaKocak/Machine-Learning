### Kısmi Türev Kavramı
Makine Öğrenmesi modellerinde genellikle birden fazla parametre vardır. Birden fazla değişkenli bir fonksiyonda "türev" almak istediğimizde, her değişkene göre ayrı ayrı türev alırız, buna kısmi türev (partial derivative) denir. Kısmi türev, "Diğer her şeyi sabit tutarsam, sadece w₁ ağırlığını değiştirmek modelin hatasını nasıl etkiler?" sorusuna cevap verir. Modeli eğitirken kullanılan Gradient Descent algoritması tamamen kısmi türevlere dayanır. 
- Her bir parametre için ayrı ayrı kısmi türev hesaplanır.
- Bu türevlerin oluşturduğu vektörlere Gradient (Gradyan) denir.
- Gradyan, hatayı en hızlı artıran yönü gösterir. Biz hatayı azaltmak istediğimiz için bu yönün tersine parametreler güncellenir.

### Örnek
- f(x, y) = 3x² + 5y

- Bu fonksiyonun x'e göre kısmi türevi alınırken, y sabit bir sayı gibi davranılır (görmezden gelinir, türevi 0 kabul edilir)
- ∂f/∂x = 6x   (y sabit kabul edildiği için türevi yok)

- Aynı şekilde y'ye göre:
- ∂f/∂y = 5   (x sabit kabul edildiği için türevi yok)

- İki kısmi türevi birleştirirsek, buna gradyan vektörü denir:
- ∇f = [∂f/∂x, ∂f/∂y] = [6x, 5]

### Chain Rule
Chain rule, iç içe geçmiş (bileşke) fonksiyonların türevini almak için kullanılan matematiksel bir kuraldır.
### Kural
Dış fonksiyonun türevi × İç fonksiyonun türevi
### Örnek
- f(x) = (2x+1)³
- 2x+1 = u olsun. f(x) = u³, Dış Türev 3u², İç Türev 2.
- Formül = 3u² * 2 = 3(2x+1)² * 2 = 6(4x²+4x+1) = 24x²+24x+6
