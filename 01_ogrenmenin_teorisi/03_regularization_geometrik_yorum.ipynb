### Ridge ve Lasso Regresyona ait Hata Fonksiyonlarının Formülleri
- Ridge (L2):  Loss = Hata + λ × Σ(w²) 
- Lasso (L1):  Loss = Hata + λ × Σ|w|

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Çemberi (L2 - Ridge) oluştur
theta = np.linspace(0, 2*np.pi, 100)
cember_w1 = np.cos(theta)
cember_w2 = np.sin(theta)

# 2. Elmasın (L1 - Lasso) köşe noktalarını tanımla
elmas_w1 = [1, 0, -1, 0, 1]
elmas_w2 = [0, 1, 0, -1, 0]

# 3. İki şekli yan yana çiz
plt.figure(figsize=(12, 5))

# Sol grafik: Ridge (L2)
plt.subplot(1, 2, 1)
plt.plot(cember_w1, cember_w2, color='blue', linewidth=2)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.xlabel('w1')
plt.ylabel('w2')
plt.title('Ridge (L2) - Kısıtlama Bölgesi: Çember')
plt.grid(True, alpha=0.3)
plt.axis('equal')

# Sağ grafik: Lasso (L1)
plt.subplot(1, 2, 2)
plt.plot(elmas_w1, elmas_w2, color='red', linewidth=2)
plt.scatter([1, 0, -1, 0], [0, 1, 0, -1], color='black', s=80, zorder=5, label='Köşeler (w1 veya w2 = 0)')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.xlabel('w1')
plt.ylabel('w2')
plt.title('Lasso (L1) - Kısıtlama Bölgesi: Elmas')
plt.legend()
plt.grid(True, alpha=0.3)
plt.axis('equal')

plt.tight_layout()
plt.show()
```

### Yorum
Lasso Regresyon'da elmasın köşeleri olmasından dolayı değme noktaları buraya değip ağırlıkları 0'a eşitleyebilirken, Ridge Regresyon'da çemberin köşesi olmadığı için değme noktası genelde eksenler üzerine denk gelmiyor, bu yüzden ağırlıklar küçülüyor ama nadiren tam sıfır oluyor.
