### Bias-Variance Tradeoff Nedir?
Bir modelin toplam beklenen hatası aşağıdaki şekilde ayrıştırılır.
- Toplam Hata = Bias² + Variance + Irreducible Error

### Bias Nedir?
Bias (Yanlılık): Modelin ortalama tahmininin, gerçek değerden ne kadar saptığının ölçüsüdür. Modelin, verideki örüntüleri yeterince yakalayıp yakalayamadığını gösterir.

### Varyans Nedir?
Varyans, modeli farklı eğitim veri örnekleriyle eğitseydik, tahminlerin birbirinden ne kadar farklı/tutarsız çıkacağının ölçüsüdür. Test verisiyle ilgili değildir — modelin kendi farklı versiyonlarının (farklı eğitim setleriyle eğitilmiş halleri) birbirleriyle ne kadar tutarlı olduğunu gösterir.

### Irreducible Error Nedir?
Irreducible Error: Veride doğal olarak var olan, hiçbir modelin ortadan kaldıramayacağı gürültüdür.

### Bias Hesaplanması
Modelin yaptığı tüm tahminlerin ortalaması alınır ve bu değerden gerçek değer çıkarılır. Bu sayı, modelin ortalama tahmininin gerçek değerden ne kadar saptığını ölçer.

### Varyans Hesaplanması
Varyans, modelin ürettiği her bir tahminin, kendi ortalama tahmininden ne kadar uzakta dalgalandığını ölçer. Yani farklı eğitim setleriyle eğitilmiş modellerin tahminlerinin kendi içindeki tutarlılığıdır.

### Dart Tahtası Benzetmesi
- Düşük Bias + Düşük Variance → Oklar hedefin tam ortasında, birbirine yakın (ideal durum)
- Yüksek Bias + Düşük Variance → Oklar birbirine yakın ama hedeften uzak bir köşede kümelenmiş (tutarlı ama sistematik hata)
- Düşük Bias + Yüksek Variance → Oklar ortalamada hedefe yakın ama kendi aralarında dağınık (tutarsız ama ortalaması doğru)
- Yüksek Bias + Yüksek Variance → Oklar hem hedeften uzak hem kendi aralarında dağınık (en kötü senaryo)

### Önemli
Yüksek Bias, Düşük Variance → Underfitting (Model örüntüleri yakalayamamış)
Düşük Bias, Yüksek Variance → Overfitting (Model eğitim verisindeki en küçük dalgalanmaları bile ezberliyor, ezberci bir yaklaşıma sahip olunca karşısına çıkan yeni verilerde (test) başarısız oluyor.)
