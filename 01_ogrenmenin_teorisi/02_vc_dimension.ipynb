### VC Dimension Nedir? 
Model ne kadar karmaşıksa (esnekse), o kadar çok/zor veriyi ayırabilir ama o kadar da ezberlemeye (overfitting'e) yatkın olur. VC Dimension, bu "karmaşıklık/esneklik" seviyesini sayısal olarak ifade eden teorik bir kavramdır (örneğin Polynomial Regression'da derece arttıkça VC Dimension da artar).
